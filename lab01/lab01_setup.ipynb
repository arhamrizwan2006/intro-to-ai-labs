## 1. Background

**Anaconda** is a free, open-source distribution of Python (and R) built for data science and
machine learning. It bundles the core scientific-Python stack (NumPy, pandas, matplotlib,
scikit-learn) and `conda`, a package-and-environment manager that lets several projects with
conflicting dependencies live side by side on the same machine without clashing.

**Jupyter Notebook** is a browser-based interactive environment where Python code is written and
run in individual *cells*. A single notebook interleaves runnable code, its output, and
formatted explanation (Markdown, tables, plots), which is why it has become the standard format
for exploratory data analysis, reporting and teaching: work is reproducible, every step is
visible, and results sit right next to the code that produced them.

Together with **Git/GitHub** for version control, this is the toolchain used for every lab in
this course:


## 3. Git and Version Control

Git tracks every change as a *commit* -- a safe, restorable checkpoint -- so nothing is ever
permanently lost. A **branch** is a separate line of development off the stable `main` trunk,
used to try something (a new feature, a fix) without risking the working version; once the work
is good, the branch is *merged* back into `main`. GitHub stores this history online, backs it
up, and makes it shareable as a portfolio.

**Exercise 3.** Using `describe()`, identify the column with the largest ratio of standard deviation to mean, and explain in two sentences why that matters for a model that has not been given scaled inputs.

````python
# --- Exercise 3: coefficient of variation -----------------------------------
# std / mean tells you how spread out a column is RELATIVE to its own size.
# A model without scaling treats a large-variance column as more important.

numeric = df.select_dtypes(include="number")
cv = (numeric.std() / numeric.mean()).abs().sort_values(ascending=False)

print("Coefficient of variation (std / mean), largest first:\n")
print(cv.round(3))
print()
print("Largest:", cv.index[0])
````

**My two-sentence answer to Exercise 3:**

`parch` has the largest coefficient of variation (2.112), meaning its values swing the most relative to its own average. Many models that take raw numeric input (for example k-NN, SVMs or gradient descent-based regressors) compute distances or gradients directly from feature magnitude, so without scaling, `fare` and `parch` would dominate the distance calculation purely because of their units, while a well-behaved feature like `pclass` would be effectively ignored regardless of its real predictive value.
